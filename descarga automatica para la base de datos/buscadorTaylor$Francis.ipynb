```python
# Extraer el texto del span
span_text = WebDriverWait(driver, 5).until(
    EC.presence_of_element_located((By.XPATH, '/html/body/div[1]/div/div[3]/div/div/div/div/div/div[2]/div/div/div/div/div[2]/div/div[1]/div/div/div[2]/div[5]/ul/li[1]/a/span')))\
    .text


num_iteraciones = int(span_text.replace(',', ''))  


num_paginas = math.ceil(num_iteraciones / 50) 

print(f"Texto extraído: {num_paginas}")

```

```python

for i in range(1, num_paginas + 1):


    WebDriverWait(driver, 5)\
    .until(EC.element_to_be_clickable((By.CSS_SELECTOR, 'div.checkbox-container')))\
    .click()

    WebDriverWait(driver, 5)\
    .until(EC.element_to_be_clickable((By.CSS_SELECTOR, 'a.download-citations')))\
    .click()

    WebDriverWait(driver, 5)\
    .until(EC.element_to_be_clickable((By.XPATH, '/html/body/div[8]/div/div/div[4]/label[2]')))\
    .click()

    WebDriverWait(driver, 5)\
    .until(EC.element_to_be_clickable((By.CSS_SELECTOR, 'a#btn-download-citations')))\
    .click()
    
    

    print(f"Iteración {i}")

    if i < num_paginas:
        driver.get(f'https://www-tandfonline-com.crai.referencistas.com/action/doSearch?AllField=%E2%80%9Ccomputational+thinking%E2%80%9D&pageSize=50&subjectTitle=&startPage={i}')

    
    time.sleep(1)

    
```
