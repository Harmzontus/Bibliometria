```python
# Extraer el texto del span
span_text = WebDriverWait(driver, 5).until(
    EC.presence_of_element_located((By.XPATH, '/html/body/div[1]/div/div/div/div/div/div/section/div/div[2]/div[1]/div[1]/h1/span')))\
    .text


num_iteraciones = int(span_text.replace(',', '').split()[0])  


num_paginas = math.ceil(num_iteraciones / 100) 

print(f"Texto extraído: {num_paginas}")

```

```python

for i in range(1, num_paginas + 1):


    WebDriverWait(driver, 5)\
    .until(EC.element_to_be_clickable((By.CSS_SELECTOR, 'div.checkbox.SelectAllCheckbox')))\
    .click()

    WebDriverWait(driver, 5)\
    .until(EC.element_to_be_clickable((By.CSS_SELECTOR, 'button.button-link.export-all-link-button.button-link-primary.button-link-icon-left')))\
    .click()

    WebDriverWait(driver, 5)\
    .until(EC.element_to_be_clickable((By.XPATH, '/html/body/div[5]/div/div/div/p/div/div/button[4]/span/span')))\
    .click()
    

    print(f"Iteración {i}")

    if i < num_paginas:
        driver.get(f'https://www-sciencedirect-com.crai.referencistas.com/search?qs=%E2%80%9Ccomputational%20thinking%E2%80%9D&show=100&offset={i}00')

    
    time.sleep(1)

    
```
